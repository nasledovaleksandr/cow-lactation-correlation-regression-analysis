```r
df <- read.csv('/content/dataCow1.csv')
head(df)
```

```r
cor(df[,c(3,4,6)])
```

```r
df$`Порода` <- as.factor(df$`Порода`)
df$`Способ.кормления` <- as.factor(df$`Способ.кормления`)
```

```r
fit <- lm(Годовой.удой.в.кг ~ Процент.жирности, data = df)
summary(fit)
```

```r
fit <- lm(Годовой.удой.в.кг ~ Живая.масса, data = df)
summary(fit)
```

```r
fit <- lm(Годовой.удой.в.кг ~ Способ.кормления, data = df)
summary(fit)
```

```r
fit <- lm(Годовой.удой.в.кг ~ Живая.масса + Процент.жирности + Способ.кормления, data = df)
summary(fit)
```
